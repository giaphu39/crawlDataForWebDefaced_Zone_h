```python
import os
import csv

HTML_DIR = "html_pages"
CSV_FILE = "deface_labels_train.csv"

def rebuild_csv_from_html():
    files = os.listdir(HTML_DIR)

    with open(CSV_FILE, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["HTML_File_Name", "Label"])

        count = 0

        for file in files:
            if file.endswith(".html"):
                writer.writerow([file, "unlabeled"])
                count += 1

    print(f"Rebuilt CSV with {count} entries")

if __name__ == "__main__":
    rebuild_csv_from_html()
```
