```python
import os
from urllib.parse import urlparse

# Tên file đầu vào (lấy từ bước kiểm tra ALIVE của bạn)
INPUT_FILE = "urls_alive.txt"

# Thư mục chứa các file danh sách URL sau khi phân loại
OUTPUT_DIR = "Classified_URLs"

def setup_output_dir():
    """Tạo thư mục chính để chứa các file txt phân loại"""
    if not os.path.exists(OUTPUT_DIR):
        os.makedirs(OUTPUT_DIR)
        print(f"[*] Đã tạo thư mục lưu trữ: ./{OUTPUT_DIR}/")

def get_file_category(url):
    """Phân loại URL dựa trên đuôi file"""
    # Lấy phần đường dẫn của URL và tách lấy đuôi (extension)
    path = urlparse(url).path
    ext = os.path.splitext(path)[1].lower()
    
    if ext in ['.jpg', '.jpeg', '.png', '.gif', '.ico', '.svg']:
        return 'images'
    elif ext in ['.txt']:
        return 'txt'
    elif ext in ['.html', '.htm', '.php']:
        return 'web'
    else:
        return 'others' # Dành cho các link không rõ đuôi

def main():
    print("--- BẮT ĐẦU PHÂN LOẠI URL ---")
    setup_output_dir()
    
    # 1. Đọc file urls_alive.txt
    if not os.path.exists(INPUT_FILE):
        print(f"[!] Không tìm thấy file: {INPUT_FILE}. Vui lòng chạy bước check alive trước.")
        return
        
    with open(INPUT_FILE, "r", encoding="utf-8") as f:
        urls = [line.strip() for line in f if line.strip()]
        
    if not urls:
        print("[!] File trống, không có URL nào để phân loại.")
        return

    print(f"[*] Đang xử lý {len(urls)} URLs từ file {INPUT_FILE}...")

    # 2. Gom nhóm các URL
    categorized_urls = {
        'images': [],
        'txt': [],
        'web': [],
        'others': []
    }

    for url in urls:
        category = get_file_category(url)
        categorized_urls[category].append(url)

    # 3. Lưu kết quả ra các file text tương ứng trong thư mục Classified_URLs
    for category, url_list in categorized_urls.items():
        if url_list: # Chỉ tạo file nếu có link thuộc nhóm đó
            filename = f"urls_{category}.txt"
            filepath = os.path.join(OUTPUT_DIR, filename)
            
            with open(filepath, "w", encoding="utf-8") as f:
                for u in url_list:
                    f.write(u + "\n")
                    
            print(f"[+] Đã phân loại {len(url_list)} link vào -> {OUTPUT_DIR}/{filename}")

    print("\n--- HOÀN TẤT ---")

if __name__ == "__main__":
    main()
```
